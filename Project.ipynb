```python
import os
import sys

# 1. Initialize an in-memory database with starter books
library = [
    {"title": "The Hobbit", "author": "J.R.R. Tolkien", "genre": "Fantasy", "status": "Available"},
    {"title": "1984", "author": "George Orwell", "genre": "Dystopian", "status": "Checked Out"},
    {"title": "The Great Gatsby", "author": "F. Scott Fitzgerald", "genre": "Classic", "status": "Available"},
    {"title": "To Kill a Mockingbird", "author": "Harper Lee", "genre": "Classic", "status": "Checked Out"},
    {"title": "The Catcher in the Rye", "author": "J.D. Salinger", "genre": "Coming-of-Age", "status": "Available"},
    {"title": "The Lord of the Rings", "author": "J.R.R. Tolkien", "genre": "Fantasy", "status": "Checked Out"},
]

def clear_screen():
    """Clears the terminal screen for a clean, application-like layout."""
    os.system('cls' if os.name == 'nt' else 'clear')

def show_catalog(search_query=""):
    """Displays the library book catalog in a structured terminal grid."""
    clear_screen()
    print("=" * 65)
    print(f"{'📖 COMMUNITY LIBRARY CATALOG':^65}")
    print("=" * 65)
    
    # Filter books if a search query is active
    filtered_books = library
    if search_query:
        print(f"Active Search Filter: '{search_query}'\n")
        filtered_books = [
            b for b in library 
            if search_query.lower() in b["title"].lower() or search_query.lower() in b["author"].lower()
        ]
        
    if not filtered_books:
        print(" [!] No books found matching that criteria.")
    else:
        # Print structured column headers
        print(f"{'ID':<4} | {'Title':<22} | {'Author':<18} | {'Status':<12}")
        print("-" * 65)
        for idx, book in enumerate(filtered_books, start=1):
            print(f"{idx:<4} | {book['title'][:22]:<22} | {book['author'][:18]:<18} | {book['status']:<12}")
            
    print("=" * 65)

def add_book():
    """Prompt user to add a new book to the library array."""
    clear_screen()
    print("=" * 65)
    print(f"{'➕ ADD NEW BOOK TO INVENTORY':^65}")
    print("=" * 65)
    
    title = input("Enter Book Title: ").strip()
    author = input("Enter Author Name: ").strip()
    genre = input("Enter Genre (optional): ").strip() or "General"
    
    if title and author:
        library.append({"title": title, "author": author, "genre": genre, "status": "Available"})
        print(f"\n[✓] Success: '{title}' added to inventory!")
    else:
        print("\n[X] Error: Title and Author cannot be empty.")
    input("\nPress Enter to return to main menu...")

def manage_inventory():
    """Toggle a book's borrowing status or remove it entirely."""
    show_catalog()
    print("\n[⚙️ INVENTORY MANAGEMENT]")
    try:
        choice = int(input("Enter Book ID to manage (or '0' to cancel): "))
        if choice == 0 or choice > len(library):
            return
            
        selected_book = library[choice - 1]
        print(f"\nSelected: '{selected_book['title']}'")
        print("1. Toggle Status (Available / Checked Out)")
        print("2. Delete Book From Library")
        
        action = input("Choose action (1-2): ").strip()
        if action == "1":
            current = selected_book["status"]
            selected_book["status"] = "Checked Out" if current == "Available" else "Available"
            print(f"[✓] Status updated to: {selected_book['status']}")
        elif action == "2":
            library.pop(choice - 1)
            print("[✓] Book removed from catalog.")
        
        input("\nPress Enter to update display...")
    except (ValueError, IndexError):
        print("\n[X] Invalid Book ID selection.")
        input("\nPress Enter to retry...")

def main_menu():
    """The central application loop runner."""
    while True:
        show_catalog()
        print("\n [1] Search Catalog   [2] Add New Book   [3] Manage Book (Status/Delete)   [4] Exit")
        choice = input("\nSelect an option (1-4): ").strip()
        
        if choice == "1":
            query = input("Enter search term (Title or Author): ").strip()
            show_catalog(search_query=query)
            input("\nPress Enter to clear search filter...")
        elif choice == "2":
            add_book()
        elif choice == "3":
            manage_inventory()
        elif choice == "4":
            clear_screen()
            print("\nGoodbye! Thank you for visiting the library.\n")
            sys.exit()

if __name__ == "__main__":
    main_menu()
```
